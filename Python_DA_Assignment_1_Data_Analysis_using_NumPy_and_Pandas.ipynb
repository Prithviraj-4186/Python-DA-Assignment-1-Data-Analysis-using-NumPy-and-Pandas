# Numpy Array Operations:

Scenario: You are given a list of daily average temperatures (in degrees Celsius)
recorded over two weeks. Your task is to use NumPy to analyze this data.
Step wise Instructions:

1. Create a 1D NumPy Array:
Create a 1D NumPy array named temperatures_w1 for Week 1 with the
following values:
[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9]

```python
import numpy as np
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
```

2. Inspection and Properties:

● Inspect the shape, data type, and the number of elements in the array.
Inspect the shape of the temperatures_w1 array.

```python
shape = temperatures_w1.shape
data_type = temperatures_w1.dtype
num_elements = temperatures_w1.size

print("Shape:", shape)
print("Data Type:", data_type)
print("Number of Elements:", num_elements)
```

3. Array Operations:

● Convert the temperatures_w1 from Celsius to Fahrenheit using the
formula: Fahrenheit = (Celsius * 9/5) + 32

```python
temperatures_w1_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:", temperatures_w1_f)
```

● Find the maximum, minimum, and mean temperatures for the week.

```python
max_temp = np.max(temperatures_w1)
min_temp = np.min(temperatures_w1)
mean_temp = np.mean(temperatures_w1)

print(f"Maximum Temperature: {max_temp}°C")
print(f"Minimum Temperature: {min_temp}°C")
print(f"Mean Temperature: {mean_temp:.2f}°C")
```

4. Array Slicing and Indexing:

● Extract the temperatures for the first three days of the week.

```python
first_three_days = temperatures_w1[:3]

print("First 3 days temperatures:", first_three_days)
```

● Extract the temperatures for the weekend (last two days)

```python
weekend_temperatures = temperatures_w1[-2:]

print("Weekend temperatures:", weekend_temperatures)
```

● Extract the temperatures for the middle three days of the week.

```python
middle_three_days = temperatures_w1[2:5]

print("Middle 3 days temperatures:", middle_three_days)
```

5. Create a 2D Array:

Create a 2D NumPy array named temperatures where each row represents

temperatures for a different week:

★ Row 1: [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9] (Week 1)

★ Row 2: [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1] (Week 2)

6. Inspect and Slice the 2D Array:

● Inspect and print the shape, data type, and total number of elements in the
temperatures array.

```python
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Elements:", temperatures.size)
```

● Extract the temperatures for each week and the weekends (last two days)
for both weeks.

```python
print(temparatures[0])
```

# Pandas Series:

1. Creating Pandas Series:

● Create a Pandas Series named marks with the following data:

➢ Marks: 95, 92, 89, 85, 80

➢ Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'

```python
import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])

print(marks)
```

2. Indexing and Slicing:

● Use the integer index position to access the mark of the 1st rank student.

```python
rank1_mark = marks.iloc[0]

print("Mark of 1st rank student:", rank1_mark)
```

● Use the loc accessor to retrieve the marks of the top 3 ranks by
specifying their index labels.

```python
top_3_marks = marks.loc['Rank1':'Rank3']

print(top_3_marks)
```

● Use the iloc accessor to retrieve the mark of the 3rd rank student.

```python
rank3_mark = marks.iloc[2]

print("Mark of 3rd rank student:", rank3_mark)
```

● Apply a boolean mask to filter the Series and retrieve the ranks where the
marks are greater than 90.

```python
print(marks[marks>90])
```

3. Manipulating Series:

● Modify the mark of the student with the 1st rank to 100.

```python
marks['Rank1'] = 100

print(marks)
```

● Remove the entry corresponding to the last rank from the Series.

```python
marks = marks.drop('Rank5')

print(marks)
```

● Compute the CGPA by dividing each mark in the Series by 10.

```python
cgpa = marks / 10

print(cgpa)
```

# Notes:

1. Creating a Series:

Import Pandas using import pandas as pd.
Create a Series using pd.Series.

2. Indexing and Slicing:

Integer Position Indexing: Use .iloc[0] to access the 1st rank mark and .iloc[2] for the 3rd rank mark based on 0-indexed integer positions.

Label-Based Indexing:

Use .loc['Rank1':'Rank3'] to retrieve the top 3 ranks by specifying their index labels.

Boolean Masking: Filter values using marks[marks > 90] to extract entries matching the condition.

3. Manipulating the Series:

Modifying Values: Assign a new value directly using the index label.

Removing Entries: Use .drop to remove the specified rank entry.

Vectorized Operations: Compute CGPA across the entire Series using marks / 10.



# Pandas DataFrame:

1. Creating Pandas DataFrame:

● Create a Pandas DataFrame named transactions with the following data:

➢ TransactionID: 101, 102, 103, 104, 105, 106, 107, 108, 109, 110

➢ ProductCategory: 'Electronics', 'Clothing', 'Electronics', 'Furniture',
'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'

➢ Region: 'North', 'South', 'North', 'East', 'West', 'North', 'East', 'West',
'South', 'North'

➢ Amount: 200, 150, 300, 450, 200, 250, 300, 180, 350, 400

```python
import pandas as pd

transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics' ],
    'Region': ['North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]})

print(transactions)
```

2. Data Exploration:

● Display the transactions DataFrame and its basic information, including
head, tail, shape, column names, and data types.

```python
# Display first 5 rows
print("\nHead:")
print(transactions.head())
```

```python
# Display last 5 rows
print("\nTail:")
print(transactions.tail())
```

```python
# Display shape (rows, columns)
print("\nShape:")
print(transactions.shape)
```

```python
# Display column names
print("\nColumn Names:")
print(transactions.columns)
```

```python
# Display complete information
print("\nBasic Information:")
print(transactions.info())
```

● Display only the 'ProductCategory' and 'Amount' columns.

```python
print(transactions[['ProductCategory', 'Amount']])
```

● Retrieve the last 3 columns of the DataFrame using iloc or loc.

```python
print(transactions.iloc[:, -3:])
```

● Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200.

```python
filtered_transactions = transactions[(transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)]

print(filtered_transactions)
```

● Find the value counts for the 'ProductCategory' column.

```python
print(transactions['ProductCategory'].value_counts())
```

● Find the unique values in the 'Region' column.

```python
unique_regions = transactions['Region'].unique().tolist()
print(unique_regions)
```

Group by 'Region' and find the mean amount for each region.

```python
mean_amount_by_region = transactions.groupby('Region')['Amount'].mean()

print(mean_amount_by_region)
```

3. Manipulating the DataFrame:

● Modify the 'Amount' for the transaction with TransactionID 102 to 165.

```python
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print(transactions)
```

● Add a new column 'Discount' by calculating 10% of the 'Amount'.

```python
transactions['Discount'] = transactions['Amount'] * 0.10
print(transactions)
```

● Remove the row with TransactionID 109.

```python
transactions=transactions.drop(transactions[transactions['TransactionID']==109].index)
print(transactions)
```

● Delete the 'Discount' column from the DataFrame.

```python
transactions=transactions.drop(columns=['Discount'])
print(transactions)
```
